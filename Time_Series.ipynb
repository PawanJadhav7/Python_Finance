```python
#Rolling Statistics
date = "Date"
symbol = 'AAPL.O'
data_symbol = pd.DataFrame(eod_dat_csv[symbol]).dropna()
```

```python
window = 20
data_symbol['Date'] = eod_dat_csv[date]
data_symbol['min'] = data_symbol[symbol].rolling(window=window).min()
data_symbol['max'] = data_symbol[symbol].rolling(window=window).max()
data_symbol['std'] = data_symbol[symbol].rolling(window=window).std()
data_symbol['median'] = data_symbol[symbol].rolling(window=window).median()
data_symbol['mean'] = data_symbol[symbol].rolling(window=window).mean()
data_symbol['ewma'] = data_symbol[symbol].ewm(halflife=0.5,min_periods=window).mean()
data_symbol.dropna()
```

```python
ax= data_symbol[['min','mean','max']].iloc[-200:].plot(
    figsize=(10,6),
    style=['g--', 'r--', 'g--'],
    linewidth=0.8,
    legend=True,
    title='Rolling Statistics'
);
plt.show()
```

```python
#TECHNICAL ANALYSIS
data_symbol['SMA1'] = data_symbol[symbol].rolling(window=42).mean()
data_symbol['SMA2'] = data_symbol[symbol].rolling(window=252).mean()
data_symbol[[symbol,'SMA1','SMA2']].tail()
```

```python
data_symbol[[symbol,'SMA1','SMA2']].plot(figsize=(10,6));
plt.show()
```

```python
data_symbol.dropna(inplace=True)
data_symbol['positions'] = np.where(data_symbol['SMA1'] > data_symbol['SMA2'],1,-1)
ax = data_symbol[[symbol,'SMA1','SMA2','positions']].plot(figsize=(10,6),secondary_y='positions');
ax.get_legend().set_bbox_to_anchor((0.25,0.85));
plt.show()
```
