```python
# Drop the Date column before mean
eod_dat_csv_no_dates = eod_dat_csv.drop(columns=["Date"]).mean()
```

```python
eod_dat_csv_no_dates.aggregate([min,np.mean,np.std,np.median ,max]).round(2)
```

```python
eod_dat_csv.drop(columns=["Date"]).pct_change().mean().plot(kind='bar', figsize=(10,6))
plt.show()
```

```python
eod_dat_csv_dates_removed = eod_dat_csv.drop(columns=["Date"])
rets = np.log(eod_dat_csv_dates_removed / eod_dat_csv_dates_removed.shift(1))
rets.head().round(3)
```

```python
rets.cumsum().apply(np.exp).plot(figsize=(10, 6))
plt.show()
```
